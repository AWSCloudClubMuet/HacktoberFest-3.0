# Milestone 4: Diagnosing Learning Curves

![Milestone 4](milestone4.png)

# Case A :
     Train 62%, Val 60%, both flat
     Diagnosis: High bias (underfitting).
     It simply fits poorly. Flat curves mean more data adds no information the model can use. The model’s capacity or features are the bottleneck.

# Interventions:

    Increase model capacity (deeper or wider network, more trees, lower regularization).
    Better features or representation (feature engineering, interaction terms, pretrained embeddings). The current input may not contain usable signal in a form the model can exploit.
    Reduce regularization or train longer. Check that the optimizer has converged and that the learning rate isn’t stalling training.
    Do not collect more data. The flat curves show it will not help.

# Case B
     Train 99%, Val 78%, val rises slowly, gap shrinks slowly
     Diagnosis: High variance (overfitting), but data-limited.
     near-perfect training accuracy means the model is memorizing. 
     The key detail is that the gap is shrinking, so the model is not fundamentally mismatched to the task. It is starved for data relative to its capacity.

# Interventions:

    Collect more data. This is the one case where it clearly pays off, since the curves are converging.
    Data augmentation, a cheaper way to get effective extra samples.
    Regularization (weight decay, dropout, early stopping, pruning), which limits the model’s ability to memorize.
    Reduce capacity or use feature selection if more data is unavailable.
    Transfer learning or pretraining, which brings in outside information to substitute for missing labeled data

# Case C
    Train 85%, Val 91%, both rise slightly

    Diagnosis: Not a classic fit problem. Validation exceeds training, which is itself a red flag to investigate. Normally train ≥ val. 


# Plausible causes:

    Regularization applied only at training time (dropout, augmentation, noise). Training accuracy is measured with these active, while validation runs clean. This is benign and common.
    
    Validation set is easier or differently distributed (smaller, cleaner, class-imbalanced in a favorable way).
    
    Data leakage between train and validation, or label noise concentrated in the training set.
    
    Training accuracy averaged over an epoch while the model is still improving, whereas validation is measured at the end.

    The slight joint rise suggests the model is still mildly underfit and improving.

# Interventions:

    Audit first: evaluate training accuracy in eval mode (no dropout or augmentation), check for leakage and duplicates, and compare the class and difficulty distributions of both splits.
    
    Use cross-validation or a fresh held-out test set to get a trustworthy generalization estimate.
    
    Then train longer or increase capacity if the corrected numbers still show underfitting.

# Case D:
    Train 88%, Val 70%, gap stays even with 4× data

    If the gap doesn’t shrink with more data, then “not enough data” is not the problem. Something else is causing it.

# plausible explanations:

    Distribution shift. The validation data looks different from the training data (different source, time period, or population). The model learns patterns that work on training data but don’t transfer, so more of the same training data doesn’t help.

    Overfitting from a model that is too big. The model has so much capacity that it keeps memorizing, even with 4× the data.
    noisy or inconsistent labels

# One experiment to tell them apart:

    Mix the training and validation data together and split it again randomly into a new train set and new validation set.
    Train on the new train set. Test on both the new validation set and the original validation set.
